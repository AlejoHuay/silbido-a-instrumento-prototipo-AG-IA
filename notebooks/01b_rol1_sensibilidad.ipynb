## 1. Carga de Datos y Audio de Referencia

```python
ruta_csv = Path("../datos/procesados/melodia_01.csv") if Path("../datos").exists() else Path("datos/procesados/melodia_01.csv")
ruta_ref = Path("../datos/audio/referencias/flauta_referencia_01.wav") if Path("../datos").exists() else Path("datos/audio/referencias/flauta_referencia_01.wav")

notas, duraciones = cargar_melodia_csv(ruta_csv)
ref_audio, sr = sf.read(ruta_ref)
print(f"Melodía cargada: {len(notas)} eventos, duración total: {np.sum(duraciones):.3f} s")
print(f"Audio referencia: {len(ref_audio)} muestras ({len(ref_audio)/sr:.3f} s a {sr} Hz)")
```
